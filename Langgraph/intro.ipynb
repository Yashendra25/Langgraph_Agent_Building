```python
from pydantic import BaseModel, Field
```

```python
class Person(BaseModel):
    name: str = Field(..., description="The person's name")
    age: int = Field(..., description="The person's age")
    city: str = Field(..., description="The city where the person lives")
person=Person(name="yashendra", age=20, city="jaipur")
print(person)
```

```python
person1=Person(name="yashendra", age=20, city=17)
print(person1)
#this is the benfit of using pydantic it will give error if we pass wrong data type to the field.
```

```python
#model with optional fields
from typing import Optional

class Employee(BaseModel):
    name: str = Field(..., description="The employee's name")
    age: int = Field(..., description="The employee's age")
    department: Optional[str] = Field(None, description="The department where the employee works")
    salary: Optional[float] = Field(None, description="The employee's salary")
    is_active: Optional[bool] = True
employee1 = Employee(name="John Doe", age=30, department="Engineering", salary=75000, is_active=True)
print(employee1)
employee2 = Employee(name="Jane Smith", age=25)
print(employee2)
```

```python
class Classroom(BaseModel):
    class_name: str = Field(..., description="The name of the classroom")
    students: list[Person] = Field(..., description="List of students in the classroom")
    teacher: Optional[str] = Field(None, description="The name of the teacher")
    room_number:Optional[int] = Field(None, description="The room number of the classroom",strict=True)
classroom = Classroom(
    class_name="Math 101",
    students=[Person(name="Alice", age=20, city="New York"), Person(name="Bob", age=22, city="Los Angeles")],
    teacher="Mr. Smith",
    room_number=101
)
print(classroom)

try :
    classroom1 = Classroom(
        class_name="Science 101",
        students=[Person(name="Charlie", age=21, city="Chicago"), Person(name="David", age=23, city="Houston")],
        teacher="Ms. Johnson",
        room_number="102"  # This will raise a validation error because room_number should be an int
    )
    print(classroom1)
except ValueError as e:
    print(f"Error occurred: {e}")
```
